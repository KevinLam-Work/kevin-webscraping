```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.chrome.options import Options
import time

# Set up Chrome options
chrome_options = Options()
#chrome_options.add_argument("--headless")  # Run in headless mode for efficiency
chrome_options.add_argument("--disable-gpu")
chrome_options.add_argument("--no-sandbox")
chrome_options.add_argument("--disable-dev-shm-usage")

# Initialize WebDriver (assuming ChromeDriver is in PATH or specify path)
driver = webdriver.Chrome(options=chrome_options)

try:
    # Navigate to the webpage
    driver.get("https://tradingeconomics.com/australia/labour-costs")
    
    # Initialize WebDriverWait
    wait = WebDriverWait(driver, 20)  # Increased timeout to 20 seconds
    
    # Step 1: Click the "10Y" button
    try:
        ten_year_button = wait.until(EC.element_to_be_clickable((By.XPATH, '//*[@id="dateSpansDiv"]/a[3]')))
        ten_year_button.click()
        time.sleep(1)  # Allow chart to update
        print("Clicked '10Y' button")
    except Exception as e:
        print(f"Error clicking '10Y' button: {e}")
        raise

    # Step 2: Open the chart type dropdown
    try:
        chart_type_dropdown = wait.until(EC.element_to_be_clickable((By.XPATH, '//*[@id="chart"]/div/div/div[1]/div/div[3]/div/button')))
        chart_type_dropdown.click()
        time.sleep(0.5)  # Allow dropdown to open
        print("Opened chart type dropdown")
    except Exception as e:
        print(f"Error opening chart type dropdown: {e}")
        raise

    # Step 3: Select the "Line" chart type
    try:
        line_chart_button = wait.until(EC.element_to_be_clickable((By.XPATH, '//*[@id="chart"]/div/div/div[1]/div/div[3]/div/div/div[5]/button')))
        line_chart_button.click()
        time.sleep(1)  # Allow chart to update
        print("Selected 'Line' chart type")
    except Exception as e:
        print(f"Error selecting 'Line' chart type: {e}")
        raise

    # Step 4: Extract the SVG path
    try:
        svg_path_element = wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, 'path.highcharts-graph')))
        svg_path = svg_path_element.get_attribute('d')
        print("Extracted SVG Path:")
        print(svg_path)
    except Exception as e:
        print(f"Error extracting SVG path: {e}")
        raise

finally:
    # Close the browser
    driver.quit()
```
